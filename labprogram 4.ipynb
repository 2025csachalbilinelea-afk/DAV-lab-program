```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Area": [1200, 1500, np.nan, 1800, 2000, 2200, np.nan, 2500, 1600, np.nan],
    "Price": [45, 60, 55, np.nan, 95, 110, np.nan, 130, 50, 75],
    "Location": ["Mysuru", "Hassan", np.nan, "Mysuru", "Bengaluru", "Mysuru", "Bengaluru", np.nan, "Hassan", "Mysuru"]
})

print("Original Dataset:")
print(df)

print("\nMissing Values in Each Column:")
print(df.isnull().sum())

df_dropped = df.dropna()

print("\nDataset After Dropping Rows:")
print(df_dropped)

df["Area"] = df["Area"].fillna(df["Area"].mean())

print("\nArea After Mean Imputation:")
print(df["Area"])

df["Price"] = df["Price"].fillna(df["Price"].median())

print("\nPrice After Median Imputation:")
print(df["Price"])

df["Location"] = df["Location"].fillna(df["Location"].mode()[0])

print("\nLocation After Mode Imputation:")
print(df["Location"])

readings = pd.Series([28, 29, np.nan, np.nan, 30, 31, np.nan, 33])

print("\nReadings Before Forward Fill:")
print(readings)

readings = readings.ffill()

print("\nReadings After Forward Fill:")
print(readings)

print("\nFinal Dataset After Imputation:")
print(df)

print("\nMissing Values After Imputation:")
print(df.isnull().sum())
```
