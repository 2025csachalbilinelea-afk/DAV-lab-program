```python
import numpy as np
import pandas as pd

# NumPy operations
arr = np.array([15, 82, 64, 91, 43])
print("array:", arr)
print("mean:", arr.mean())
print("reshaped 2D:\n", arr.reshape(5, 1))
print("sliced(index 1 to 3):", arr[1:4])

# Pandas operations
data = {
    "Name": ['Priya', 'Kavya', 'Rohan', 'Suraj'],
    "Marks": [65, 88, 72, 94],
    "Branch": ['ECE', 'ECE', 'ECE', 'ECE'],
}

df = pd.DataFrame(data)
print("\n First two rows:\n", df.iloc[0:2])
print("\n Marks column\n", df['Marks'])
print("\n Rows where Marks>70:\n", df[df['Marks'] > 70])
```
